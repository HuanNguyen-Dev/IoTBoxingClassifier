```python
labelled_data = []

for _, row in raw_df.iterrows():
    for label_timeline in row['label_timeline']:
        label = label_timeline['label']
        start = label_timeline['start']
        end = label_timeline['end']

        start_ms = start * 1000
        end_ms = end * 1000

        labelled_row = {'label': label, 'participant': row.participant, 'record_id': row.name,}

        for suffix in SUFFIXES:
            device_times = np.asarray(row[f't_dev_ms{suffix}'])

            # Find indices corresponding to start/end timestamps
            start_idx = np.searchsorted(device_times, start_ms, side='left')
            end_idx = np.searchsorted(device_times, end_ms, side='left')

            for channel in CHANNELS:
                labelled_row[f'{channel}{suffix}'] = (
                    row[f'{channel}{suffix}'][start_idx:end_idx]
                )

        labelled_data.append(labelled_row)

labelled_df = pd.DataFrame(labelled_data)
```

```python
def IIR_filter(value, prev, alpha):
    return alpha * value + (1 - alpha) * prev

filtered_data = []
alpha = SAMPLING_PERIOD / (0.05 + SAMPLING_PERIOD)

for _, row in labelled_df.iterrows():
    new_row = {'label': row.label, 'participant': row.participant, 'record_id': row.name}
    for channel in CHANNELS:
        for suffix in SUFFIXES:
            data = row[f"{channel}{suffix}"].copy()

            prev = 0
            for i, record in enumerate(data):
                data[i] = IIR_filter(record, prev, alpha)
                prev = record
            
            new_row[f"{channel}{suffix}"] = data

    filtered_data.append(new_row)

filtered_df = pd.DataFrame(filtered_data)
```

## Extract Features

```python
def extract_features(signal):
    signal = np.asarray(signal, dtype=float)
    keys = ['mean','std','min','max','range','median','rms',
            'skew','kurtosis','iqr','p10','p90','energy',
            'argmax_pos','argmin_pos','zero_cross','mean_abs_jerk',
            'q1_mean','q2_mean','q3_mean','q4_mean']

    if len(signal) < 4:
        return {k: np.nan for k in FEATURES}

    centred = signal - signal.mean()
    jerk = np.diff(signal)
    q = np.array_split(signal, 4)
    n = len(signal)

    return {
        'mean': signal.mean(),
        'std': signal.std(),
        'min': signal.min(),
        'max': signal.max(),
        'range': np.ptp(signal),
        'median': np.median(signal),
        'rms': np.sqrt(np.mean(signal**2)),
        'skew': skew(signal),
        'kurtosis': kurtosis(signal),
        'iqr': np.percentile(signal, 75) - np.percentile(signal, 25),
        'p10': np.percentile(signal, 10),
        'p90': np.percentile(signal, 90),
        'energy': np.sum(signal**2),
        'argmax_pos': np.argmax(signal) / n,
        'argmin_pos': np.argmin(signal) / n,
        'zero_cross': np.sum(np.diff(np.sign(centred)) != 0),
        'mean_abs_jerk': np.mean(np.abs(jerk)),
        'q1_mean': q[0].mean(), 'q2_mean': q[1].mean(),
        'q3_mean': q[2].mean(), 'q4_mean': q[3].mean(),
    }
```

```python
feature_data = []

for _, row in filtered_df.iterrows():

    feature_row = {
        'record_id': row.record_id,
        'participant': row.participant,
        'label': row.label
    }

    for suffix in SUFFIXES:
        for channel in CHANNELS:
            signal = row[f'{channel}{suffix}']

            features = extract_features(signal)

            for feature_name, value in features.items():
                feature_row[
                    f'{channel}{suffix}_{feature_name}'
                ] = value

    feature_data.append(feature_row)

features_df = pd.DataFrame(feature_data)
```

## Augment Data

```python
rng = np.random.default_rng(42)

def augment_data(raw_df, n_copies=5):
    rows = []
    for _, row in raw_df.iterrows():
        for _ in range(n_copies):
            feat = {'record_id': row.record_id, 'participant': row.participant, 'label': row.label}
            for suffix in SUFFIXES:
                scale_factor = rng.uniform(0.8, 1.2)                     # one scale factor per sensor
                for channel in CHANNELS:
                    x = np.asarray(row[f'{channel}{suffix}'], float)
                    x = x * scale_factor + rng.normal(0, 0.05 * (x.std() + 1e-9), len(x))
                    for k, v in extract_features(x).items():
                        feat[f'{channel}{suffix}_{k}'] = v
            rows.append(feat)
    return pd.DataFrame(rows)
```

# Training and Validation

```python
def lopo_hand(suffix, state, data):
    columns = HAND_FEATURE_COLUMNS[suffix]
    participants = data["participant"].unique()
    if len(participants) < 2:
        raise ValueError("LOPO needs at least two participants")
    rows = []
    for participant in participants:
        train = data[data.participant != participant]
        test = data[data.participant == participant]
        if set(train["label"]) != set(HAND_LABELS[suffix]):
            raise ValueError(f"{suffix}: missing training classes when holding out {participant}")
        model = RandomForestClassifier(
            n_estimators=300, class_weight="balanced", random_state=state
        )
        model.fit(train[columns], train["label"])
        pred = model.predict(test[columns])
        punch = test["label"].to_numpy() != "rest"
        if not punch.any():
            raise ValueError(f"{suffix}: no punches to evaluate for {participant}")
        rows.append({
            "held_out": participant,
            "acc": accuracy_score(test["label"], pred),
            "acc_punches_only": accuracy_score(test.loc[punch, "label"], pred[punch]),
            "f1_punches_only": f1_score(
                test.loc[punch, "label"], pred[punch],
                labels=[label for label in HAND_LABELS[suffix] if label != "rest"],
                average="macro", zero_division=0,
            ),
        })
    return pd.DataFrame(rows).set_index("held_out")
```

## Leave-One-Participant-Out (LOPO) Cross-Validation

```python
feat_cols = [c for c in features_df.columns if c not in ['label', 'participant', 'record_id']]

#normalise data with respect to each participant
def per_user_norm(df):
    out = df.copy()
    for participant, row in df.groupby('participant'):
        out.loc[row.index, feat_cols] = (row[feat_cols] - row[feat_cols].mean()) / (row[feat_cols].std() + 1e-8)
    return out

# Leave-One-Participant-Out (LOPO) cross-validation
real_norm = per_user_norm(features_df)
aug_norm  = per_user_norm(aug_df)

def lopo(model, use_aug=True):
    rows = []
    for p in real_norm.participant.unique():
        train = real_norm[real_norm.participant != p]
        if use_aug:
            train = pd.concat([train, aug_norm[aug_norm.participant != p]])
        test = real_norm[real_norm.participant == p]          # real data only
        model.fit(train[feat_cols], train['label'])
        pred = model.predict(test[feat_cols])
        punch = test.label != le.transform(['rest'])[0]
        rows.append({'held_out': p,
                     'acc': accuracy_score(test.label, pred),
                     'acc_punches_only': accuracy_score(test.label[punch], pred[punch]),
                     'f1_punches_only': f1_score(test.label[punch], pred[punch], average='macro')})
    return pd.DataFrame(rows).set_index('held_out')

```
